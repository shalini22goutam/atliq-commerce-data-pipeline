```python
import yaml

table_name = dbutils.widgets.get("table_name")

if not table_name:
    raise ValueError("Missing required parameter: table_name")

CONFIG_PATH = "/Workspace/Users/shalini22goutam@gmail.com/atliq-commerce-data-pipeline/databricks_jobs/src/notebooks/check_schema_drift"

with open(CONFIG_PATH, "r") as f:
    schema_baseline = yaml.safe_load(f)
if table_name not in schema_baseline:
    raise ValueError(f"No schema baseline configured for table='{table_name}'")

actual_columns = set(spark.table(table_name).columns)
expected_columns = set(schema_baseline[table_name])

missing = expected_columns - actual_columns
extra = actual_columns - expected_columns

if missing or extra:
    raise ValueError(f"Schema mismatch for {table_name}: missing={missing}, extra={extra}")

dbutils.notebook.exit("Schema check passed")
```
